```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

def manual_shift_outliers(x, y, shift_value=0):
    """
    Интерактивная функция для выбора и сдвига выбросов
    """
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(x, y, 'b-', picker=True, pickradius=5)
    
    selected_points = []
    
    def on_pick(event):
        index = event.ind[0]
        if index not in selected_points:
            selected_points.append(index)
            ax.plot(x[index], y[index], 'ro', markersize=10)
            fig.canvas.draw()
            print(f"Выбрана точка {index}: x={x[index]:.2f}, y={y[index]:.2f}")
    
    fig.canvas.mpl_connect('pick_event', on_pick)
    plt.show()
    
    # После выбора точек применяем сдвиг
    response = input("Введите индексы для сдвига (через запятую) или 'q' для выхода: ")
    if response.lower() != 'q':
        indices = [int(i.strip()) for i in response.split(',')]
        shift = float(input(f"Введите величину сдвига: "))
        
        y_new = y.copy()
        y_new[indices] += shift
        
        # Показываем результат
        plt.figure(figsize=(10, 4))
        plt.plot(x, y, 'b-', alpha=0.5, label='Исходные')
        plt.plot(x, y_new, 'g-', label='После сдвига')
        plt.plot(x[indices], y_new[indices], 'ro', label='Сдвинутые точки')
        plt.legend()
        plt.grid(True)
        plt.show()
        
        return y_new
    
    return y

# Использование
# y_corrected = manual_shift_outliers(x, y)
```
